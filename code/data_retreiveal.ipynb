```python
import yfinance as yf
import pandas as pd
from fredapi import Fred
import os
```

```python
# Retrieving FRED data to get 10Y Treasury Yield, Consumer Price Index and Fed Funds effective rates
FRED_API_KEY = os.environ.get("FRED_API_KEY")
FRED_SERIES = {
    "DGS10": "10Y Treasury Yield (%)",
    "CPIAUCSL": "CPI (index, not seasonally adj. headline)",
    "FEDFUNDS": "Effective Fed Funds Rate (%)",
}

START_DATE = "2000-01-01"
END_DATE = "2025-01-01"


def fetch_macro_data():
    print("Fetching macro data from FRED...")
    fred = Fred(api_key=FRED_API_KEY)
    macro_frames = {}
    for series_id, label in FRED_SERIES.items():
        print(f"  - {series_id}: {label}")
        macro_frames[series_id] = fred.get_series(
            series_id, observation_start=START_DATE, observation_end=END_DATE
        )
    macro = pd.DataFrame(macro_frames)
    macro.index.name = "date"

    return macro
```
